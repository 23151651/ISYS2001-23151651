## Test before you submit

Run your tracker with the scenario below and confirm the figures match. Then test it against awkward input: a word where a number is expected, a negative amount, and `done` entered immediately with no expenses. It should handle all of these without crashing.


```python

# Starting values
total_spent = 0
expense_count = 0
largest_expense = 0


while True:

    while True:
        try:
            expense = float(input("Enter an expense amount: $"))

            if expense < 0:
                print("Error: Can't put in negatives.")
            else:
                break

        except ValueError:
            print("Error: Please enter an actual number.")

    total_spent += expense

    expense_count += 1

    if expense > largest_expense:
        largest_expense = expense

    print("Expense added!")

    while True:
        choice = input(
            "Any more expenses? (yes/no): "
        ).lower()

        if choice == "yes":
            break

        elif choice == "no":
            break

        else:
            print("What? Please enter 'yes' or 'no'.")

    if choice == "no":
        break


average_expense = total_spent / expense_count


# ---------------------------------------
#      Spending Summary
# ---------------------------------------

print("\n" + "=" * 40)
print("           SPENDING SUMMARY")
print("=" * 40)

print(f"Number of expenses: {expense_count}")
print(f"Total spent:        ${total_spent:.2f}")
print(f"Average expense:    ${average_expense:.2f}")
print(f"Largest expense:    ${largest_expense:.2f}")

print("=" * 40)

# Final message
print("Thank you for using the Finance Tracker!")
```
